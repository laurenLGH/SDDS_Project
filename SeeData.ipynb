## Set up environment

1) Create Virtual Environment
In terminal, run ```python -m venv .venv```

2) Activate the virtual environment
In terminal, run ```source .venv/bin/activate```

3) Install required packaged into virtual environment
In terminal, run ```pip install -r requirements.txt```

4) Make venv available for notebook
```python -m ipykernel install --user --name=SDDS_Project```
